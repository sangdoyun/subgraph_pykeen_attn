```python
from rdflib import Graph, Namespace, URIRef, Literal
from rdflib.namespace import RDF, XSD
from datetime import datetime, timedelta
import random
import hashlib

# Namespaces
ex = Namespace("http://example.org/")
health = Namespace("http://example.org/health/")
finance = Namespace("http://example.org/finance/")
general = Namespace("http://example.org/general/")
time = Namespace("http://example.org/time/")

g = Graph()
g.bind("ex", ex)
g.bind("health", health)
g.bind("finance", finance)
g.bind("general", general)
g.bind("time", time)

def generate_stmt_id(s, p, o):
    raw = f"{str(s)}_{str(p)}_{str(o)}"
    return URIRef(f"http://example.org/event/{hashlib.md5(raw.encode()).hexdigest()}")

def add_temporal_triple(g, s, p, o, timestamp):
    g.add((s, p, o))
    #stmt_id = generate_stmt_id(s, p, o)
    #g.add((stmt_id, time.timestamp, Literal(timestamp.isoformat(), datatype=XSD.dateTime)))
    g.add((o, time.timestamp, Literal(timestamp.isoformat(), datatype=XSD.dateTime)))

# --- Template-driven Event Simulations ---
def simulate_doctor_visit(g, person, doctor, location, symptoms, diagnoses, start_date):
    event_time = start_date + timedelta(days=random.randint(0, 5))
    for symptom in symptoms:
        add_temporal_triple(g, person, health.hasSymptom, health[symptom], event_time)
    for diagnosis in diagnoses:
        add_temporal_triple(g, person, health.diagnosedWith, health[diagnosis], event_time)
    add_temporal_triple(g, person, health.visitedDoctor, health[doctor], event_time)
    add_temporal_triple(g, person, health.atHospital, health[location], event_time)

def simulate_finance_event(g, person, bank, asset, amount, start_date):
    event_time = start_date + timedelta(days=random.randint(0, 5))
    add_temporal_triple(g, person, finance.investedIn, finance[asset], event_time)
    add_temporal_triple(g, person, finance.accountAt, finance[bank], event_time)
    add_temporal_triple(g, person, finance.transactionAmount, Literal(amount, datatype=XSD.float), event_time)

def simulate_routine_activity(g, person, activity, location, start_date):
    event_time = start_date + timedelta(days=random.randint(0, 5))
    add_temporal_triple(g, person, general.participatedIn, general[activity], event_time)
    add_temporal_triple(g, person, general.locatedAt, general[location], event_time)

# --- Drift Injection ---
def inject_drift(g, person, start_date):
    drift_day = random.randint(10, 20)
    d = start_date + timedelta(days=drift_day)
    # Health drift: new condition or increased visits
    simulate_doctor_visit(g, person, "DrBob", "CityClinic", ["Fatigue"], ["Diabetes"], d)
    # Routine drift: change activity/location
    simulate_routine_activity(g, person, "Yoga", "WellnessCenter", d)
    # Travel drift: simulate relocation
    add_temporal_triple(g, person, general.relocatedTo, general["NewCity"], d)

# --- Simulate for multiple people ---
people = [ex.JohnDoe, ex.JaneSmith, ex.Robert]
start_date = datetime(2025, 1, 1)

for i in range(60):  # simulate for 30 days
    for person in people:
        d = start_date + timedelta(days=i)

        if random.random() < 0.4:
            simulate_doctor_visit(
                g, person, "DrAlice", "CityClinic",
                symptoms=["Cough", "Fever"],
                diagnoses=["Flu"],
                start_date=d
            )

        if random.random() < 0.3:
            simulate_finance_event(
                g, person, "BankA", "MutualFundX", round(random.uniform(100, 500), 2), d
            )

        if random.random() < 0.5:
            simulate_routine_activity(
                g, person, "Workout", "GymPark", d
            )

# Inject temporal drift for each person
for person in people:
    inject_drift(g, person, start_date)

```

```python
def evaluate_drift(predictions, ground_truth_index):
    predicted_drift = [
        1 if label in ('drifting', 'decaying') else 0
        for _, _, label in predictions
    ]
    true_labels = [1 if i == ground_truth_index else 0 for i in range(len(predictions))]

    TP = sum(p == t == 1 for p, t in zip(predicted_drift, true_labels))
    FP = sum(p == 1 and t == 0 for p, t in zip(predicted_drift, true_labels))
    FN = sum(p == 0 and t == 1 for p, t in zip(predicted_drift, true_labels))
    TN = sum(p == t == 0 for p, t in zip(predicted_drift, true_labels))

    precision = TP / (TP + FP + 1e-5)
    recall = TP / (TP + FN + 1e-5)
    f1 = 2 * precision * recall / (precision + recall + 1e-5)
    accuracy = (TP + TN) / len(predictions)

    try:
        delay = next(i for i, pred in enumerate(predicted_drift) if pred == 1) - ground_truth_index
    except StopIteration:
        delay = None

    return {
        'accuracy': round(accuracy, 3),
        'precision': round(precision, 3),
        'recall': round(recall, 3),
        'f1_score': round(f1, 3),
        'detection_delay': delay
    }
```

```python
import numpy as np
import networkx as nx
from sklearn.metrics.pairwise import cosine_similarity
from collections import defaultdict
from karateclub import Graph2Vec
from node2vec import Node2Vec

# === Embedding Utilities ===
def compute_graph2vec_embeddings(graphs):
    model = Graph2Vec(dimensions=64, wl_iterations=2, epochs=10)
    model.fit(graphs)
    return model.get_embedding()

def compute_node2vec_embeddings(graph):
    node2vec = Node2Vec(graph, dimensions=64, walk_length=10, num_walks=100, workers=1)
    model = node2vec.fit(window=5, min_count=1)
    return {str(n): model.wv[str(n)] for n in graph.nodes() if str(n) in model.wv}

# === Core Functions ===
def graph_cosine_sim(emb1, emb2):
    return cosine_similarity([emb1], [emb2])[0][0]

def avg_node_sim(subg1, subg2, node_emb_t, node_emb_tp):
    sims = []
    for v in subg1.nodes():
        if v in subg2.nodes() and v in node_emb_t and v in node_emb_tp:
            sim = cosine_similarity(
                [node_emb_t[v]],
                [node_emb_tp[v]]
            )[0][0]
            sims.append(sim)
        else:
            sims.append(0.0)
    return np.mean(sims) if sims else 0.0

def compute_similarity_matrices(subgraphs_t, subgraphs_tp, threshold=0.5):
    n = len(subgraphs_t)
    Sg = np.zeros((n, n))
    Sn = np.zeros((n, n))

    # Compute embeddings
    emb_t = compute_graph2vec_embeddings(subgraphs_t)
    emb_tp = compute_graph2vec_embeddings(subgraphs_tp)

    node_emb_t = {i: compute_node2vec_embeddings(subgraphs_t[i]) for i in range(n)}
    node_emb_tp = {j: compute_node2vec_embeddings(subgraphs_tp[j]) for j in range(n)}

    for i in range(n):
        for j in range(n):
            Sg[i, j] = graph_cosine_sim(emb_t[i], emb_tp[j])
            Sn[i, j] = avg_node_sim(subgraphs_t[i], subgraphs_tp[j], node_emb_t[i], node_emb_tp[j])

    return Sg, Sn

def compute_evolution_signature(Sg, Sn, threshold=0.5):
    n = Sg.shape[0]
    Sig = np.empty((n, n), dtype=object)

    for i in range(n):
        for j in range(n):
            if Sg[i, j] > threshold and Sn[i, j] > threshold:
                Sig[i, j] = "static"
            elif Sg[i, j] > threshold and Sn[i, j] <= threshold:
                Sig[i, j] = "growing"
            elif Sg[i, j] <= threshold and Sn[i, j] > threshold:
                Sig[i, j] = "possible drift"
            else:
                Sig[i, j] = "drifted"

    return Sig

# === Entry Function ===
def temporal_evolution_signature(subgraphs_t, subgraphs_tp, threshold=0.5):
    """
    subgraphs_t: List of nx.DiGraph at reference time t
    subgraphs_tp: List of nx.DiGraph at time t'
    threshold: similarity threshold for classification
    """
    Sg, Sn = compute_similarity_matrices(subgraphs_t, subgraphs_tp, threshold)
    Sig = compute_evolution_signature(Sg, Sn, threshold)
    return Sg, Sn, Sig
```

```python
# Final Code: Temporal Drift Detection with Corrected TKG and Evolution Signature

from rdflib import Graph, URIRef, Literal, Namespace
from rdflib.namespace import XSD
from datetime import datetime, timedelta
import hashlib
import random
import networkx as nx
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity
from karateclub import Graph2Vec
from node2vec import Node2Vec

# Namespaces
ex = Namespace("http://example.org/")
health = Namespace("http://example.org/health/")
finance = Namespace("http://example.org/finance/")
general = Namespace("http://example.org/general/")
time = Namespace("http://example.org/time/")

# --- TKG Construction (Object -> Time) ---
g = Graph()
g.bind("ex", ex)
g.bind("health", health)
g.bind("finance", finance)
g.bind("general", general)
g.bind("time", time)

def add_temporal_triple(g, s, p, o, timestamp):
    g.add((s, p, o))
    g.add((o, time.timestamp, Literal(timestamp.isoformat(), datatype=XSD.dateTime)))

# --- Subgraph Extraction Based on Context Namespace ---
def extract_context_subgraph(graph, person_uri, context_prefix, until_time=None):
    G = nx.DiGraph()
    for s, p, o in graph:
        if not isinstance(o, URIRef):
            continue
        if str(s) != str(person_uri):
            continue
        ts = None
        for _, _, tval in graph.triples((o, time.timestamp, None)):
            ts = datetime.fromisoformat(str(tval))
        if until_time and (not ts or ts > until_time):
            continue
        if str(p).startswith(context_prefix) or str(o).startswith(context_prefix):
            G.add_edge(str(s), str(o), label=str(p).split("/")[-1])
    return G

# --- Snapshot Extraction Using Context Subgraphs ---
def extract_weekly_subgraphs(graph, person_uri, context_prefix, window_days=7, num_snapshots=10):
    all_ts = [o for s, p, o in graph.triples((None, time.timestamp, None))]
    all_ts = sorted([datetime.fromisoformat(str(ts)) for ts in all_ts])
    start = all_ts[0] if all_ts else datetime.now()
    subgraphs = []
    for i in range(num_snapshots):
        end_time = start + timedelta(days=window_days * (i + 1))
        G = extract_context_subgraph(graph, person_uri, context_prefix, until_time=end_time)
        subgraphs.append(G)
    return subgraphs

# --- Embedding Computation ---
def compute_graph2vec_embeddings(graphs):
    model = Graph2Vec(dimensions=64, wl_iterations=2, epochs=10)
    model.fit(graphs)
    return model.get_embedding()

def compute_node2vec_embeddings(graph):
    if graph.number_of_edges() == 0 or graph.number_of_nodes() == 0:
        return {}
    node2vec = Node2Vec(graph, dimensions=64, walk_length=10, num_walks=100, workers=1)
    model = node2vec.fit(window=5, min_count=1)
    return {str(n): model.wv[str(n)] for n in graph.nodes() if str(n) in model.wv}

# --- Similarity + Signature ---
def graph_cosine_sim(emb1, emb2):
    return cosine_similarity([emb1], [emb2])[0][0]

def avg_node_sim(subg1, subg2, node_emb_t, node_emb_tp):
    sims = []
    for v in subg1.nodes():
        if v in subg2.nodes() and v in node_emb_t and v in node_emb_tp:
            sims.append(cosine_similarity([node_emb_t[v]], [node_emb_tp[v]])[0][0])
        else:
            sims.append(0.0)
    return np.mean(sims) if sims else 0.0

def compute_similarity_matrices(subgraphs_t, subgraphs_tp, threshold=0.5):
    n = len(subgraphs_t)
    Sg = np.zeros((n, n))
    Sn = np.zeros((n, n))

    emb_t = compute_graph2vec_embeddings(subgraphs_t)
    emb_tp = compute_graph2vec_embeddings(subgraphs_tp)
    node_emb_t = {i: compute_node2vec_embeddings(subgraphs_t[i]) for i in range(n)}
    node_emb_tp = {j: compute_node2vec_embeddings(subgraphs_tp[j]) for j in range(n)}

    for i in range(n):
        for j in range(n):
            Sg[i, j] = graph_cosine_sim(emb_t[i], emb_tp[j])
            Sn[i, j] = avg_node_sim(subgraphs_t[i], subgraphs_tp[j], node_emb_t[i], node_emb_tp[j])

    return Sg, Sn

def compute_evolution_signature(Sg, Sn, threshold=0.5):
    n = Sg.shape[0]
    Sig = np.empty((n, n), dtype=object)
    for i in range(n):
        for j in range(n):
            if Sg[i, j] > threshold and Sn[i, j] > threshold:
                Sig[i, j] = "static"
            elif Sg[i, j] > threshold and Sn[i, j] <= threshold:
                Sig[i, j] = "growing"
            elif Sg[i, j] <= threshold and Sn[i, j] > threshold:
                Sig[i, j] = "possible drift"
            else:
                Sig[i, j] = "drifted"
    return Sig

# --- Evaluation ---
def evaluate_drift_signature(sig_matrix, ground_truth_week):
    predictions = []
    for i in range(sig_matrix.shape[0]):
        row_labels = list(sig_matrix[i])
        has_drift = any(label in ('drifted', 'possible drift', 'growing') for label in row_labels)
        predictions.append(1 if has_drift else 0)

    true_labels = [1 if i == ground_truth_week else 0 for i in range(sig_matrix.shape[0])]

    TP = sum(p == t == 1 for p, t in zip(predictions, true_labels))
    FP = sum(p == 1 and t == 0 for p, t in zip(predictions, true_labels))
    FN = sum(p == 0 and t == 1 for p, t in zip(predictions, true_labels))
    TN = sum(p == t == 0 for p, t in zip(predictions, true_labels))

    precision = TP / (TP + FP + 1e-5)
    recall = TP / (TP + FN + 1e-5)
    f1 = 2 * precision * recall / (precision + recall + 1e-5)
    accuracy = (TP + TN) / len(predictions)

    try:
        delay = next(i for i, pred in enumerate(predictions) if pred == 1) - ground_truth_week
    except StopIteration:
        delay = None

    return {
        'accuracy': round(accuracy, 3),
        'precision': round(precision, 3),
        'recall': round(recall, 3),
        'f1_score': round(f1, 3),
        'detection_delay': delay
    }
```

```python
import networkx as nx
from node2vec import Node2Vec
import numpy as np
import matplotlib.pyplot as plt

# 1. Create a dummy graph (like a toy Knowledge Graph)
G = nx.Graph()

# Add nodes (entities)
nodes = ["Person_A", "Person_B", "Company_X", "Location_Y", "Product_Z"]
G.add_nodes_from(nodes)

# Add edges (relationships)
edges = [
    ("Person_A", "Company_X"),    # works_at
    ("Person_B", "Company_X"),    # works_at
    ("Company_X", "Location_Y"),  # located_in
    ("Person_A", "Product_Z"),    # uses
    ("Person_B", "Product_Z")     # uses
]
G.add_edges_from(edges)

# Draw the graph (optional)
# plt.figure(figsize=(6, 4))
# nx.draw(G, with_labels=True, node_color='lightblue', node_size=2000, font_size=10)
# plt.title("Dummy Knowledge Graph")
# plt.show()

# 2. Node2Vec Embedding
# Create a Node2Vec instance
node2vec = Node2Vec(G, dimensions=64, walk_length=10, num_walks=100, workers=1)

# Fit model
model = node2vec.fit(window=5, min_count=1, batch_words=4)

# Get node embeddings
node_embeddings = {node: model.wv[node] for node in G.nodes()}

# Print example embedding
print("Embedding for 'Person_A':")
print(node_embeddings["Person_A"])

# 3. Aggregate Node Embeddings to create Graph Embedding
# Simple mean-pooling for graph-level embedding
graph_embedding = np.mean(np.array(list(node_embeddings.values())), axis=0)

print("\nGraph-level embedding (mean-pooled):")
print(graph_embedding)
```
